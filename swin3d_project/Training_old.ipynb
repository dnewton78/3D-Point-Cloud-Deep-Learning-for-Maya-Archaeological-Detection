```python
import torch


def move_batch_to_device(batch, device):
    coords, features, offset, labels, paths = batch

    coords = coords.to(
        device,
        non_blocking=True,
    ).float()

    features = features.to(
        device,
        non_blocking=True,
    ).float()

    offset = offset.to(
        device,
        non_blocking=True,
    ).long()

    labels = labels.to(
        device,
        non_blocking=True,
    ).long()

    return coords, features, offset, labels


def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device,
):
    model.train()

    total_loss = 0.0
    total_correct = 0
    total_points = 0

    for batch in loader:
        coords, features, offset, labels = move_batch_to_device(
            batch,
            device,
        )

        optimizer.zero_grad(set_to_none=True)

        # Model expects:
        # (coords, features, offset)
        logits = model(
            (coords, features, offset)
        )

        # Some Swin3D implementations may return a tuple.
        if isinstance(logits, tuple):
            logits = logits[0]

        loss = criterion(logits, labels)

        loss.backward()
        optimizer.step()

        predictions = logits.argmax(dim=1)

        number_of_points = labels.numel()

        total_loss += loss.item() * number_of_points
        total_correct += (
            predictions == labels
        ).sum().item()
        total_points += number_of_points

    average_loss = total_loss / max(total_points, 1)
    accuracy = total_correct / max(total_points, 1)

    return average_loss, accuracy


@torch.no_grad()
def validate(
    model,
    loader,
    criterion,
    device,
):
    model.eval()

    total_loss = 0.0
    total_correct = 0
    total_points = 0

    for batch in loader:
        coords, features, offset, labels = move_batch_to_device(
            batch,
            device,
        )

        logits = model(
            (coords, features, offset)
        )

        if isinstance(logits, tuple):
            logits = logits[0]

        loss = criterion(logits, labels)

        predictions = logits.argmax(dim=1)

        number_of_points = labels.numel()

        total_loss += loss.item() * number_of_points
        total_correct += (
            predictions == labels
        ).sum().item()
        total_points += number_of_points

    average_loss = total_loss / max(total_points, 1)
    accuracy = total_correct / max(total_points, 1)

    return average_loss, accuracy
```

```python
import torch


def move_batch_to_device(batch, device):
    coords, features, offset, labels, paths = batch

    coords = coords.to(
        device,
        non_blocking=True,
    ).float()

    features = features.to(
        device,
        non_blocking=True,
    ).float()

    offset = offset.to(
        device,
        non_blocking=True,
    ).long()

    labels = labels.to(
        device,
        non_blocking=True,
    ).long()

    return coords, features, offset, labels


def calculate_batch_accuracy(logits, labels):
    predictions = logits.argmax(dim=1)

    correct = (
        predictions == labels
    ).sum().item()

    total = labels.numel()

    return correct, total


def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device,
):
    model.train()

    total_loss = 0.0
    total_correct = 0
    total_points = 0

    for batch_index, batch in enumerate(loader):
        coords, features, offset, labels = (
            move_batch_to_device(
                batch,
                device,
            )
        )

        optimizer.zero_grad(set_to_none=True)

        logits = model(
            (coords, features, offset)
        )

        if isinstance(logits, tuple):
            logits = logits[0]

        if logits.shape[0] != labels.shape[0]:
            raise RuntimeError(
                "Logit and label point counts do not match: "
                f"logits={logits.shape}, labels={labels.shape}"
            )

        loss = criterion(
            logits,
            labels,
        )

        if not torch.isfinite(loss):
            raise RuntimeError(
                f"Non-finite loss detected in batch {batch_index}: "
                f"{loss.item()}"
            )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0,
        )

        optimizer.step()

        correct, number_of_points = (
            calculate_batch_accuracy(
                logits,
                labels,
            )
        )

        total_loss += (
            loss.item() * number_of_points
        )

        total_correct += correct
        total_points += number_of_points

        if (
            batch_index == 0
            or (batch_index + 1) % 10 == 0
        ):
            print(
                f"  Batch {batch_index + 1:04d}/"
                f"{len(loader):04d} | "
                f"Loss: {loss.item():.6f}"
            )

        del coords
        del features
        del offset
        del labels
        del logits
        del loss

    average_loss = (
        total_loss / max(total_points, 1)
    )

    accuracy = (
        total_correct / max(total_points, 1)
    )

    return average_loss, accuracy


@torch.no_grad()
def validate(
    model,
    loader,
    criterion,
    device,
):
    model.eval()

    total_loss = 0.0
    total_correct = 0
    total_points = 0

    for batch in loader:
        coords, features, offset, labels = (
            move_batch_to_device(
                batch,
                device,
            )
        )

        logits = model(
            (coords, features, offset)
        )

        if isinstance(logits, tuple):
            logits = logits[0]

        if logits.shape[0] != labels.shape[0]:
            raise RuntimeError(
                "Logit and label point counts do not match: "
                f"logits={logits.shape}, labels={labels.shape}"
            )

        loss = criterion(
            logits,
            labels,
        )

        correct, number_of_points = (
            calculate_batch_accuracy(
                logits,
                labels,
            )
        )

        total_loss += (
            loss.item() * number_of_points
        )

        total_correct += correct
        total_points += number_of_points

    average_loss = (
        total_loss / max(total_points, 1)
    )

    accuracy = (
        total_correct / max(total_points, 1)
    )

    return average_loss, accuracy
```

```python
from models.swin3d_model import Model
from losses.generalized_dice import GeneralizedDiceLoss


model = Model(
    num_classes=TRAIN_CONFIG["num_classes"],
    in_channels=4,
).to(device)

criterion = GeneralizedDiceLoss(
    num_classes=TRAIN_CONFIG["num_classes"],
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=TRAIN_CONFIG["learning_rate"],
    weight_decay=TRAIN_CONFIG["weight_decay"],
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=TRAIN_CONFIG["max_epochs"],
    eta_min=1e-7,
)
```

```python
best_validation_loss = float("inf")
epochs_without_improvement = 0

max_epochs = TRAIN_CONFIG["max_epochs"]
patience = TRAIN_CONFIG["early_stopping_patience"]

training_history = {
    "train_loss": [],
    "train_accuracy": [],
    "validation_loss": [],
    "validation_accuracy": [],
    "learning_rate": [],
}


for epoch in range(max_epochs):
    print(
        f"\n{'=' * 60}\n"
        f"Epoch {epoch + 1}/{max_epochs}\n"
        f"{'=' * 60}"
    )

    train_loss, train_accuracy = train_one_epoch(
        model=model,
        loader=train_loader,
        criterion=criterion,
        optimizer=optimizer,
        device=device,
    )

    validation_loss, validation_accuracy = validate(
        model=model,
        loader=validation_loader,
        criterion=criterion,
        device=device,
    )

    current_learning_rate = (
        optimizer.param_groups[0]["lr"]
    )

    training_history["train_loss"].append(
        train_loss
    )

    training_history["train_accuracy"].append(
        train_accuracy
    )

    training_history["validation_loss"].append(
        validation_loss
    )

    training_history["validation_accuracy"].append(
        validation_accuracy
    )

    training_history["learning_rate"].append(
        current_learning_rate
    )

    print(
        f"\nEpoch {epoch + 1:03d} results\n"
        f"Train loss:       {train_loss:.6f}\n"
        f"Train accuracy:   {train_accuracy:.4f}\n"
        f"Validation loss:  {validation_loss:.6f}\n"
        f"Validation acc:   {validation_accuracy:.4f}\n"
        f"Learning rate:    {current_learning_rate:.8f}"
    )

    checkpoint = {
        "epoch": epoch + 1,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_state_dict": scheduler.state_dict(),
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "validation_loss": validation_loss,
        "validation_accuracy": validation_accuracy,
        "best_validation_loss": best_validation_loss,
        "training_history": training_history,
        "config": TRAIN_CONFIG,
    }

    torch.save(
        checkpoint,
        last_checkpoint_path,
    )

    if validation_loss < best_validation_loss:
        best_validation_loss = validation_loss
        epochs_without_improvement = 0

        checkpoint["best_validation_loss"] = (
            best_validation_loss
        )

        torch.save(
            checkpoint,
            best_checkpoint_path,
        )

        print(
            f"Saved new best model: "
            f"{best_checkpoint_path}"
        )

    else:
        epochs_without_improvement += 1

        print(
            "No validation improvement: "
            f"{epochs_without_improvement}/{patience}"
        )

    scheduler.step()

    if epochs_without_improvement >= patience:
        print(
            f"\nEarly stopping at epoch {epoch + 1}."
        )
        print(
            f"Best validation loss: "
            f"{best_validation_loss:.6f}"
        )
        break

    torch.cuda.empty_cache()
```
