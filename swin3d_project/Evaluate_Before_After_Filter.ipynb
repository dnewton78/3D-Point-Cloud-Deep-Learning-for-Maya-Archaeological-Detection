# Spatial Post-Processing: Before vs After

The same spatial filter used in the visualization notebook is applied to every test tile:

- 1 m XY grid
- 3×3 morphological opening
- Structure components smaller than 8 cells removed
- Plazuela components smaller than 15 cells removed
- Removed archaeological predictions are changed to Ground

The filter is applied **tile by tile** before recomputing the test metrics.


```python
# ============================================================
# Load the best checkpoint explicitly
# ============================================================

checkpoint = torch.load(
    BEST_CHECKPOINT,
    map_location=device,
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

print(
    "Loaded checkpoint epoch:",
    checkpoint.get("epoch", "unknown"),
)
```

```python
# ============================================================
# Filter settings
# ============================================================

FILTER_SETTINGS = {
    "cell_size": 1.0,
    "structure_min_cells": 8,
    "plazuela_min_cells": 15,
    "opening_iterations": 1,
    "closing_iterations": 0,
}

print(FILTER_SETTINGS)
```

```python
# ============================================================
# Helpers
# ============================================================

CLASS_NAMES = [
    "Ground",
    "Structure",
    "Plazuela",
]

NUM_CLASSES = 3


def add_to_confusion_matrix(
    confusion_matrix,
    labels,
    predictions,
):
    labels = np.asarray(
        labels,
        dtype=np.int64,
    )

    predictions = np.asarray(
        predictions,
        dtype=np.int64,
    )

    valid = (
        (labels >= 0)
        &
        (labels < NUM_CLASSES)
        &
        (predictions >= 0)
        &
        (predictions < NUM_CLASSES)
    )

    encoded = (
        NUM_CLASSES * labels[valid]
        +
        predictions[valid]
    )

    counts = np.bincount(
        encoded,
        minlength=NUM_CLASSES * NUM_CLASSES,
    )

    confusion_matrix += counts.reshape(
        NUM_CLASSES,
        NUM_CLASSES,
    )


def metrics_from_confusion_matrix(
    confusion_matrix,
):
    cm = confusion_matrix.astype(
        np.float64
    )

    total = cm.sum()

    overall_accuracy = (
        np.trace(cm) / total
        if total > 0
        else 0.0
    )

    rows = []

    for class_id, class_name in enumerate(
        CLASS_NAMES
    ):
        tp = cm[class_id, class_id]

        fn = (
            cm[class_id, :].sum()
            -
            tp
        )

        fp = (
            cm[:, class_id].sum()
            -
            tp
        )

        precision = (
            tp / (tp + fp)
            if (tp + fp) > 0
            else 0.0
        )

        recall = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else 0.0
        )

        f1 = (
            2.0
            * precision
            * recall
            / (precision + recall)
            if (precision + recall) > 0
            else 0.0
        )

        iou = (
            tp
            / (tp + fp + fn)
            if (tp + fp + fn) > 0
            else 0.0
        )

        rows.append(
            {
                "Class": class_name,
                "Precision": precision,
                "Recall": recall,
                "F1": f1,
                "IoU": iou,
                "GT Points": int(
                    cm[class_id, :].sum()
                ),
                "Predicted Points": int(
                    cm[:, class_id].sum()
                ),
                "True Positives": int(tp),
                "False Positives": int(fp),
                "False Negatives": int(fn),
            }
        )

    return overall_accuracy, pd.DataFrame(
        rows
    )
```

```python
# ============================================================
# Evaluate all test tiles BEFORE and AFTER filtering
# ============================================================

confusion_before = np.zeros(
    (NUM_CLASSES, NUM_CLASSES),
    dtype=np.int64,
)

confusion_after = np.zeros(
    (NUM_CLASSES, NUM_CLASSES),
    dtype=np.int64,
)

total_removed_structure = 0
total_removed_plazuela = 0

tiles_processed = 0


with torch.no_grad():

    for batch_index, batch in enumerate(
        test_loader
    ):

        (
            coords,
            features,
            offset,
            labels,
            paths,
            metric_coords,
        ) = batch

        coords_gpu = coords.to(
            device,
            non_blocking=True,
        ).float()

        features_gpu = features.to(
            device,
            non_blocking=True,
        ).float()

        offset_gpu = offset.to(
            device,
            non_blocking=True,
        ).long()

        logits = model(
            (
                coords_gpu,
                features_gpu,
                offset_gpu,
            )
        )

        if isinstance(logits, tuple):
            logits = logits[0]

        predictions = (
            logits.argmax(dim=1)
            .cpu()
            .numpy()
        )

        coords_np = (
            coords.cpu()
            .numpy()
        )

        labels_np = (
            labels.cpu()
            .numpy()
        )

        # ----------------------------------------------------
        # Process each tile separately
        # ----------------------------------------------------

        start = 0

        for tile_in_batch, end in enumerate(
            offset.tolist()
        ):

            tile_coords = coords_np[
                start:end
            ]

            tile_labels = labels_np[
                start:end
            ]

            tile_predictions = predictions[
                start:end
            ]

            # BEFORE
            add_to_confusion_matrix(
                confusion_before,
                tile_labels,
                tile_predictions,
            )

            # AFTER spatial filtering
            (
                filtered_predictions,
                filter_info,
            ) = spatial_filter_predictions(
                coords=tile_coords,
                predictions=tile_predictions,
                **FILTER_SETTINGS,
                verbose=False,
            )

            add_to_confusion_matrix(
                confusion_after,
                tile_labels,
                filtered_predictions,
            )

            total_removed_structure += (
                filter_info[1][
                    "removed_points"
                ]
            )

            total_removed_plazuela += (
                filter_info[2][
                    "removed_points"
                ]
            )

            tiles_processed += 1

            start = end

        if (
            batch_index == 0
            or
            (batch_index + 1) % 10 == 0
            or
            batch_index + 1 == len(test_loader)
        ):
            print(
                f"Processed test batch "
                f"{batch_index + 1}/"
                f"{len(test_loader)}"
            )


print()
print(
    "Tiles processed:",
    tiles_processed,
)

print(
    "Structure predictions removed:",
    f"{total_removed_structure:,}",
)

print(
    "Plazuela predictions removed:",
    f"{total_removed_plazuela:,}",
)
```

```python
# ============================================================
# Calculate BEFORE and AFTER metrics
# ============================================================

(
    accuracy_before,
    per_class_before,
) = metrics_from_confusion_matrix(
    confusion_before
)

(
    accuracy_after,
    per_class_after,
) = metrics_from_confusion_matrix(
    confusion_after
)


print("=" * 65)
print("OVERALL ACCURACY")
print("=" * 65)

print(
    f"Before filter: {accuracy_before:.4f}"
)

print(
    f"After filter:  {accuracy_after:.4f}"
)

print(
    f"Change:        "
    f"{accuracy_after - accuracy_before:+.4f}"
)
```
